```python
import pandas as pd
import glob
from tqdm import tqdm
```

# Change batches from zip to CBSA

```python
from cgnn.process_xwalk import get_zip_cbsa_map

zip_cbsa_map = get_zip_cbsa_map()
```

```python
indir = "/Users/hwunrow/Documents/GitHub/cgnn/data/raw/advan/2023/"
files = glob.glob(f"{indir}/batch_*.csv")

for file in tqdm(files):
    try:
        df = pd.read_csv(file, dtype={"zip_orig": str, "zip_dest": str})
        df = df.merge(
            zip_cbsa_map.rename(columns={"ZIP": "zip_orig", "CBSA": "cbsa_orig"}),
            on="zip_orig",
            how="left",
        )

        df = df.merge(
                zip_cbsa_map.rename(columns={"ZIP": "zip_dest", "CBSA": "cbsa_dest"}),
                on="zip_dest",
                how="left",
            )

        df = df.groupby(["date_range_start", "date_range_end", "cbsa_orig", "cbsa_dest"])[["visitor_home_aggregation"]].sum().reset_index()

        df.to_csv(file, index=False)
    except:
        print(f"Error reading file: {file}")
        continue
    

```

# Combines batches

```python
indir = "/Users/hwunrow/Documents/GitHub/cgnn/data/raw/advan/"

for year in tqdm(range(2020, 2026)):
    files = glob.glob(f"{indir}/{year}/batch_*.csv")
    print(year, len(files))
    df = pd.concat([pd.read_csv(file, dtype={"cbsa_orig": str, "cbsa_dest": str}) for file in files])
    print(df.info())

    df.to_csv(f"{indir}/{year}_advan.csv", index=False)


```
